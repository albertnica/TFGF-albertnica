```python
sql = f"""
SELECT TOP 10
   p.objid,
   p.ra,
   p.dec,
   p.run,
   p.rerun,
   p.camcol,
   p.field,
   s.z AS redshift
FROM PhotoObj AS p
JOIN SpecObj AS s ON s.bestobjid = p.objid
WHERE s.class = 'QSO'
"""
cj.executeQuery(sql, context='DR18')
```

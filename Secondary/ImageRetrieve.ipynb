```python
sql = """
SELECT TOP 10
   p.objid,
   p.ra,
   p.dec,
   p.run,
   p.rerun,
   p.camcol,
   p.field,
   s.z AS redshift
FROM PhotoObj AS p
JOIN SpecObj AS s ON s.bestobjid = p.objid
WHERE s.class = 'QSO'
"""
results = cj.executeQuery(sql, context='DR18')
```

```python
import os
import requests
import pandas as pd

os.makedirs("C:\\Users\\alfa\\Desktop\\DATA", exist_ok=True)
df = pd.DataFrame(results, columns=['objid', 'ra', 'dec', 'run', 'rerun', 'camcol', 'field', 'redshift'])
df[['ra', 'dec']].to_csv('C:\\Users\\alfa\\Downloads\\tfg tests\\sql tests and web retrieving\\images\\query_results.csv', index=False)
```

```python
data = pd.read_csv('C:\\Users\\alfa\\Desktop\\DATA\\query_results.csv')  # CSV with columns 'ra' and 'dec'

# Example function to fetch and save an image by RA and DEC
def fetch_image(ra, dec, scale=0.2, width=512, height=512, save_path='C:\\Users\\alfa\\Downloads\\tfg tests\\sql tests and web retrieving\\images\\'):
    url = f'https://skyserver.sdss.org/dr16/SkyServerWS/ImgCutout/getjpeg?ra={ra}&dec={dec}&scale={scale}&width={width}&height={height}'
    response = requests.get(url)
    if response.status_code == 200:
        filename = f"{save_path}image_{ra}_{dec}.jpg"
        with open(filename, 'wb') as file:
            file.write(response.content)
        print(f"Saved image for RA: {ra}, DEC: {dec}")
    else:
        print(f"Failed to retrieve image for RA: {ra}, DEC: {dec}")

# Loop through the data to fetch and save each image
for index, row in data.iterrows():
    fetch_image(row['ra'], row['dec'])
```

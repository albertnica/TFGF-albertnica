```python
import os
from astropy.io import fits

# Directorio de origen
source_dir = r'spectrums'

# Obtener y ordenar la lista de archivos .fits
files = sorted([f for f in os.listdir(source_dir) if f.endswith('.fits')])

# Definir el archivo desde el cual quieres iniciar la iteración
start_file = 'spec-2193-53888-0595.fits'

# Si el archivo existe en la lista, comenzar desde él; de lo contrario, iterar todos
if start_file in files:
    start_index = files.index(start_file)
    files = files[start_index:]
else:
    print(f"Archivo de inicio '{start_file}' no encontrado. Se procesarán todos los archivos.")

# Iterar sobre los archivos a partir del archivo de inicio
for file_name in files:
    file_path = os.path.join(source_dir, file_name)
    try:
        with fits.open(file_path) as hdul:
            if len(hdul) < 3:
                print(f"Archivo '{file_name}' no tiene suficientes HDUs")
                continue

            # Extraer la clase desde el HDU 2
            file_class = hdul[2].data["CLASS"]

            # Si es iterable pero no es una cadena, tomar el primer elemento
            if hasattr(file_class, '__iter__') and not isinstance(file_class, str):
                file_class = file_class[0]

            # Normalizar la clase a mayúsculas
            file_class_str = str(file_class).strip().upper()

        # Si la clase es "STAR", eliminar el archivo
        if file_class_str == 'STAR':
            os.remove(file_path)
            print(f"Archivo '{file_name}' eliminado")
    except Exception as e:
        print(f"Error al procesar '{file_name}': {e}")
```

```python
import os
import numpy as np
import matplotlib.pyplot as plt

# Directorio donde se encuentra el archivo memmap
dst_folder = "data"
redshift_dat_filename = "spectra_data_500k_redshift_mmap.dat"
redshift_dat_path = os.path.join(dst_folder, redshift_dat_filename)

# Número total de espectros
total_spectra = 460000

# Cargar el memmap en modo lectura
redshift_memmap = np.memmap(redshift_dat_path, dtype='float32', mode='r', shape=(total_spectra,))

# Convertir a array de NumPy
redshift_array = np.array(redshift_memmap)
print(len(redshift_array))
      
# Definir los bins para el histograma (ancho de bin = 0.1)
bins = np.arange(0, redshift_array.max() + 0.1, 0.1)

# Crear y mostrar el histograma
plt.figure(figsize=(18, 16))
plt.hist(redshift_array, bins=bins, color='skyblue', edgecolor='black')
plt.xlabel("Redshift")
plt.ylabel("Número de espectros")
plt.title("Histograma de redshifts (ancho de bin = 0.1)")
plt.xlim(0, 7.1)
plt.show()
```
